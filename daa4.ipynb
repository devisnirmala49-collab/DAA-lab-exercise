```python
import sys

# Function to find the vertex with minimum distance
def minimum_distance(distance, visited, vertices):
    minimum = sys.maxsize
    minimum_index = -1

    for vertex in range(vertices):
        if distance[vertex] < minimum and visited[vertex] == False:
            minimum = distance[vertex]
            minimum_index = vertex

    return minimum_index


# Function to implement Dijkstra's Algorithm
def dijkstra(graph, source, vertices):

    distance = [sys.maxsize] * vertices
    visited = [False] * vertices

    distance[source] = 0

    for count in range(vertices):

        u = minimum_distance(distance, visited, vertices)

        visited[u] = True

        for v in range(vertices):

            if (graph[u][v] > 0 and
                visited[v] == False and
                distance[v] > distance[u] + graph[u][v]):

                distance[v] = distance[u] + graph[u][v]

    print("\nShortest Distance from Source Vertex", source)
    print("----------------------------------------")
    print("Vertex\tDistance")

    for i in range(vertices):
        print(i, "\t", distance[i])


# ---------------- Main Program ----------------

vertices = int(input("Enter the number of vertices: "))

print("\nEnter the Adjacency Matrix:")

graph = []

for i in range(vertices):
    row = list(map(int, input().split()))
    graph.append(row)

source = int(input("\nEnter the Source Vertex: "))

dijkstra(graph, source, vertices)
```
