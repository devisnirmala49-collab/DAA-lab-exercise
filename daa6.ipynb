```python
# Matrix Chain Multiplication using Dynamic Programming

import sys

# Function to calculate minimum multiplication cost
def matrix_chain_order(p):

    n = len(p)

    # Create DP table
    m = [[0 for i in range(n)] for j in range(n)]

    # Chain length starts from 2
    for length in range(2, n):

        for i in range(1, n - length + 1):

            j = i + length - 1

            m[i][j] = sys.maxsize

            for k in range(i, j):

                cost = (m[i][k] +
                        m[k + 1][j] +
                        p[i - 1] * p[k] * p[j])

                if cost < m[i][j]:
                    m[i][j] = cost

    return m


# ---------------- Main Program ----------------

n = int(input("Enter the number of matrices: "))

print("\nEnter the dimensions:")

dimensions = []

for i in range(n + 1):
    value = int(input(f"Dimension {i + 1}: "))
    dimensions.append(value)

table = matrix_chain_order(dimensions)

print("\nDynamic Programming Cost Table:\n")

for i in range(1, n + 1):
    for j in range(1, n + 1):

        if j < i:
            print("-", end="\t")
        else:
            print(table[i][j], end="\t")

    print()

print("\nMinimum Number of Multiplications =", table[1][n])
```
