```python
# Program to Find Minimum and Maximum
# Using Divide and Conquer Technique

def findMinMax(arr, low, high):

    # If there is only one element
    if low == high:
        minimum = arr[low]
        maximum = arr[low]
        return minimum, maximum

    # If there are two elements
    elif high == low + 1:

        if arr[low] < arr[high]:
            minimum = arr[low]
            maximum = arr[high]
        else:
            minimum = arr[high]
            maximum = arr[low]

        return minimum, maximum

    # More than two elements
    else:

        mid = (low + high) // 2

        left_min, left_max = findMinMax(arr, low, mid)

        right_min, right_max = findMinMax(arr, mid + 1, high)

        # Compare minimum values
        if left_min < right_min:
            minimum = left_min
        else:
            minimum = right_min

        # Compare maximum values
        if left_max > right_max:
            maximum = left_max
        else:
            maximum = right_max

        return minimum, maximum


# ---------------- Main Program ----------------

n = int(input("Enter the number of elements: "))

arr = []

print("Enter the elements:")

for i in range(n):
    value = int(input())
    arr.append(value)

minimum, maximum = findMinMax(arr, 0, n - 1)

print("\nArray Elements:", arr)
print("Minimum Value =", minimum)
print("Maximum Value =", maximum)
```
