```python
import heapq

# -----------------------------
# Prim's Algorithm
# -----------------------------
def prim(graph):
    visited = set()
    min_heap = [(0, 0, -1)]  # (weight, vertex, parent)
    mst = []
    total_cost = 0

    while min_heap:
        weight, vertex, parent = heapq.heappop(min_heap)

        if vertex in visited:
            continue

        visited.add(vertex)

        if parent != -1:
            mst.append((parent, vertex, weight))
            total_cost += weight

        for neighbor, wt in graph[vertex]:
            if neighbor not in visited:
                heapq.heappush(min_heap, (wt, neighbor, vertex))

    return mst, total_cost


# -----------------------------
# Kruskal's Algorithm
# -----------------------------
class DisjointSet:
    def __init__(self, n):
        self.parent = list(range(n))

    def find(self, x):
        if self.parent[x] != x:
            self.parent[x] = self.find(self.parent[x])
        return self.parent[x]

    def union(self, x, y):
        px = self.find(x)
        py = self.find(y)

        if px != py:
            self.parent[py] = px
            return True
        return False


def kruskal(n, edges):
    edges.sort(key=lambda x: x[2])

    ds = DisjointSet(n)
    mst = []
    total_cost = 0

    for u, v, w in edges:
        if ds.union(u, v):
            mst.append((u, v, w))
            total_cost += w

    return mst, total_cost


# -----------------------------
# Main Program
# -----------------------------
n = int(input("Enter number of vertices: "))
e = int(input("Enter number of edges: "))

graph = [[] for _ in range(n)]
edges = []

print("Enter edges (u v weight):")
for _ in range(e):
    u, v, w = map(int, input().split())
    graph[u].append((v, w))
    graph[v].append((u, w))
    edges.append((u, v, w))

# Prim's Algorithm
prim_mst, prim_cost = prim(graph)

print("\nPrim's Minimum Spanning Tree:")
for u, v, w in prim_mst:
    print(f"{u} -- {v} = {w}")
print("Total Cost:", prim_cost)

# Kruskal's Algorithm
kruskal_mst, kruskal_cost = kruskal(n, edges)

print("\nKruskal's Minimum Spanning Tree:")
for u, v, w in kruskal_mst:
    print(f"{u} -- {v} = {w}")
print("Total Cost:", kruskal_cost)
```
