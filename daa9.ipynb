```python
import math

def tsp(graph, n):
    visited = [False] * n
    visited[0] = True

    min_cost = math.inf
    best_path = []

    def branch(path, cost):
        nonlocal min_cost, best_path

        current = path[-1]

        # All cities visited
        if len(path) == n:
            total_cost = cost + graph[current][0]

            if total_cost < min_cost:
                min_cost = total_cost
                best_path = path[:] + [0]

            return

        # Try every unvisited city
        for city in range(n):
            if not visited[city]:

                new_cost = cost + graph[current][city]

                # Branch and Bound
                if new_cost < min_cost:
                    visited[city] = True
                    path.append(city)

                    branch(path, new_cost)

                    path.pop()
                    visited[city] = False

    branch([0], 0)

    return best_path, min_cost


# Cost matrix
graph = [
    [0, 10, 15, 20],
    [10, 0, 35, 25],
    [15, 35, 0, 30],
    [20, 25, 30, 0]
]

n = len(graph)

path, cost = tsp(graph, n)

print("Optimal Path:", " -> ".join(map(str, path)))
print("Minimum Cost:", cost)
```
